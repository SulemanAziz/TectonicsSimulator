```python
import pandas as pd;

polygons = pd.read_json("../MerdithStaticPolygons.json");
polygons = pd.read_json((polygons.polygons).to_json());
polygons = polygons.transpose();
polygons.to_csv("Polygons.csv");
```

```python
polygondata = pd.read_csv("Polygons.csv", usecols=["name", "featureType", "plateId", "points"]);
polygondata["points"] = polygondata["points"].astype(str).str.replace(r'[^0-9,.{}-]', '', regex=True);
polygondata.to_csv("polygonsfiltered.csv");
```

```python
platerotations = pd.read_json("../MerdithPlateRotations.json");
platerotations = pd.read_json((platerotations.records).to_json())
platerotations = platerotations.transpose();
platerotations.to_csv("platerotations.csv");
```

```python
topologyboundaries = pd.read_json("../MerdithTopologyBoundaries.json");
topologyboundaries = pd.read_json((topologyboundaries.boundaries).to_json());
topologyboundaries = topologyboundaries.transpose();
topologyboundaries.to_csv("topologyboundaries.csv");

topocsv = pd.read_csv("topologyboundaries.csv");
topocsv = topocsv.drop(columns=["sourceFile"]);
topocsv["points"] = topocsv["points"].astype(str).str.replace(r'[^0-9,.{}-]', '', regex=True);
topocsv.to_csv("topologyboundariesfiltered.csv")
```
